```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

PROJECT_ROOT = next(
    p for p in [Path.cwd().resolve(), *Path.cwd().resolve().parents]
    if (p / 'date').is_dir() and (p / 'images').is_dir()
)
DATA_DIR = PROJECT_ROOT / 'date'
IMAGES_DIR = PROJECT_ROOT / 'images'
sns.set_theme(style="whitegrid",font_scale = 1.2)

plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
```

```python
#数据读取
df = pd.read_parquet(DATA_DIR / 'user_behavior_cleaned_dataset.parquet')
print(df.head(5))
```

```python
result_pv = df[df['behavior_type'] == 'pv']
result_intent = df[df['behavior_type'].isin(['fav', 'cart'])]
result_buy = df[df['behavior_type'] == 'buy']

pv_users = set(result_pv['user_id'])
first_item_pv_time = (
    result_pv.groupby(['user_id', 'item_id'])['timestamp']
    .min()
    .rename('first_item_pv_time')
)

# 意向阶段：同一用户对同一商品先浏览，再收藏或加购。
intent_result = result_intent.merge(
    first_item_pv_time,
    on=['user_id', 'item_id'],
    how='inner'
)
intent_result = intent_result[
    intent_result['timestamp'] >= intent_result['first_item_pv_time']
]
first_item_intent_time = (
    intent_result.groupby(['user_id', 'item_id'])['timestamp']
    .min()
    .rename('first_item_intent_time')
)

# 支付阶段：同一商品的收藏或加购时间不晚于购买时间。
buy_result = result_buy.merge(
    first_item_pv_time,
    on=['user_id', 'item_id'],
    how='inner'
)
buy_result = buy_result[buy_result['timestamp'] >= buy_result['first_item_pv_time']]
buy_result = buy_result.merge(
    first_item_intent_time,
    on=['user_id', 'item_id'],
    how='inner'
)
buy_result = buy_result[buy_result['timestamp'] >= buy_result['first_item_intent_time']]

count_pv = len(pv_users)
count_intent = intent_result['user_id'].nunique()
count_buy = buy_result['user_id'].nunique()

print(f"浏览用户{count_pv}，同一商品浏览后收藏或加购{count_intent}人，"
      f"同一商品完成浏览到支付{count_buy}人；"
      f"浏览到意向转化率{count_intent/count_pv:.2%}，浏览到支付转化率{count_buy/count_pv:.2%}")
```

```python
df_ratio = pd.DataFrame({
      '行为':['pv','fav/cart','buy'],
      '人数':[count_pv,count_intent,count_buy]
})
fig,axes = plt.subplots(figsize=(14,5))
colors = plt.cm.Blues(np.linspace(0.8,0.3,len(df_ratio)))
axes.bar(df_ratio['行为'],df_ratio['人数'],color = colors,width=0.4)
for i,(behavior,count) in enumerate(zip(df_ratio['行为'],df_ratio['人数'])):
    axes.text(i,count + max(df_ratio['人数']) * 0.02,
              f'{count:.0f}',
              ha = 'center',
              va = 'bottom',
              fontsize = 12,
              fontweight = 'bold',
              )
axes.set_title('同商品用户转化漏斗',fontsize=14)
axes.set_xlabel('行为类型',fontsize=12)
axes.set_ylabel('人数',fontsize=12)
axes.set_ylim([0,max(df_ratio['人数'])*1.25])

sns.despine()

plt.tight_layout()
plt.savefig(IMAGES_DIR / 'traffic_funnel_analysis.png',dpi=300)
```
